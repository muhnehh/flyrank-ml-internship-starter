```python
import pandas as pd

# Load the anonymized starter dataset from the repository
df = pd.read_csv('/content/content_refresh_anonymized.csv')

# 1. Scope: Total pages we are analyzing
total_pages = len(df)

# 2. Opportunity: Pages with high visibility but poor click-through
high_imp_low_ctr = len(df[(df['impressions_90d'] >= 1000) & (df['ctr'] < 0.02)])

# 3. Context: What is the average position of these underperforming pages?
avg_pos = df[(df['impressions_90d'] >= 1000) & (df['ctr'] < 0.02)]['avg_position'].mean()

print(f"Dataset scope: {total_pages:,} total pages.")
print(f"Immediate Opportunity Size: {high_imp_low_ctr:,} pages have >1,000 recent impressions but a CTR under 2%.")
print(f"Average rank of these missed opportunities: Position {avg_pos:.1f}.")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

* **What I can claim:**
  * **Observed Performance Gaps:** In our 30,000-page dataset, we identified 1,235 high-visibility pages (>1,000 recent impressions) with underperforming CTRs (<2%) sitting at an average position of 25.2.
  * **Directional Decision-Support:** This work provides a position- and volume-adjusted opportunity queue to help editorial teams focus resources where the gap between expected and actual performance is largest.

* **What I CANNOT claim:**

  * **Algorithm Reverse-Engineering:** This model does not "predict Google's algorithm" or account for unobserved SERP layout dynamics (e.g., featured snippets, local packs, or ad placement variations).
